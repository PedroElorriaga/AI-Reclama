```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_comments = pd.read_csv(r'E:\Projetos-Codes\AI-ReclamaEm\data\processed\reclamacoes_processado.csv')
```

```python
comments_count = df_comments['sentimento'].value_counts()
comments_count
```

```python
xy = comments_count.plot(kind='bar', title='Distribuição de sentimentos', ylabel='Número de Reclamações')
xy.bar_label(xy.containers[0])
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
from wordcloud import WordCloud
```

```python
negative_texts = ' '.join(df_comments[df_comments['sentimento'] == 'NEGATIVO']['texto_limpo'])
cloud = WordCloud(width=2000, height=1000, background_color='white').generate(negative_texts)

plt.imshow(cloud, interpolation='bilinear')
plt.axis('off') # SEM MARCADORES
plt.title('Palavras frequentes em reclamações')
plt.show()
```
